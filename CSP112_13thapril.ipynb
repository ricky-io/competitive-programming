# Q1: Student Performance Analysis:

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

marks = np.random.randint(0, 101, size=(100, 5))
student_ids = np.arange(1, 101)
mean_marks = marks.mean(axis=1)

highest = marks.max()
lowest = marks.min()

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

```python
df = pd.DataFrame(marks, columns=['Subject1', 'Subject2', 'Subject3', 'Subject4', 'Subject5'])
df.insert(0, 'Student_ID', student_ids)
df['Average'] = mean_marks

def get_grade(avg):
    if avg > 80:
        return 'A'
    elif avg >= 60:
        return 'B'
    else:
        return 'C'

df['Grade'] = df['Average'].apply(get_grade)

print(df.head())
```

```python
plt.figure()
plt.hist(df['Average'], bins=10)
plt.title("Histogram of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
grade_counts = df['Grade'].value_counts()

plt.figure()
grade_counts.plot(kind='bar')
plt.title("Number of Students per Grade")
plt.xlabel("Grade")
plt.ylabel("Count")
plt.show()
```

```python
## i. 'C' grade has the heighest frequency.
## ii. Bell shape curve; most student fell in average case.
```

# Q2: Sales Data Trend Analysis:

```python
days = 60
dates = pd.date_range(start="2026-02-01", periods=days)
sales = np.random.randint(1000, 10000, size=days)
noise = np.random.normal(0, 500, size=days)
sales = sales + noise
df_sales = pd.DataFrame({
    'Date': dates,
    'Sales': sales
})
df_sales['Moving_Avg'] = df_sales['Sales'].rolling(window=7).mean()
df_sales['Growth_Rate'] = df_sales['Sales'].pct_change() * 100
print(df_sales)
```

```python
plt.figure()
plt.plot(df_sales['Date'], df_sales['Sales'], label='Daily Sales')
plt.plot(df_sales['Date'], df_sales['Moving_Avg'], label='7-Day Moving Avg')
peak_day = df_sales.loc[df_sales['Sales'].idxmax()]
plt.scatter(peak_day['Date'], peak_day['Sales'])
plt.title("Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
# i. highly fluctuations
# ii. high volatile
# iii. should be focus on sales on peak days
```
